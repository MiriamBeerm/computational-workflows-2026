### What is a container? 
A container is a loosely isolated environment in which applictaions can be packaged and run. Containers contain everything needed to run the application, so it is not necessaryto rely on what's installed on the host. Containers can be shared while working, while everyone gets the same container that works in the same way.

### Why do we use containers? 
Te pipelines we use rely on dozens of specialized bioinformatics tools each needing specific versions.
Instead of installing all those complex dependencies directly on the computer, Docker containers come with everything pre-installed and pre-configured. Nextflow automatically downloads the correct container for each step, runs the analysis safely inside it, and then closes it down. Moreover the sharing of containers enhance the reproducibility of worklows.

### What is a docker image?
A Docker image is a standalone and read-only software package that contains everything needed to run an application. It serves as a static template used to running containers.

### Now that you know how to use a docker container, which approach was easier and which approach will be easier in the future? (eig wäre Vergleich mit noraml fastqc download ohne docker mit fastqc docker mit seqera containers)
For manual tests it is unimportant which approach is choosed, but for the future for nextflow-workflows it is easier to use seqera containers, because nextflow is using Seqera/Wave-infrastructure. Traditional containers (such as those from Quay.io) are rigid, pre-built packages. Seqera Wave, on the other hand, can dynamically assemble or optimize container layers based on demand. This saves storage space and often resolves compatibility issues.

### What would you say, which approach is more reproducible? (eig wäre Vergleich mit noraml fastqc download ohne docker mit fastqc docker mit seqera containers)
When it comes to pure, standalone static code locking, BioContainers (Quay.io) has a slight advantage, since the image is finalized the moment it is created.
When it comes to executing modern workflows and flexibility in hardware selection, Seqera Wave offers the same level of reproducibility while also solving the problem of compatibility between different computers.
