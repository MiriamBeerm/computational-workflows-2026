1. What is nf-core?

nf-core is a global community that works with the Nextflow workflow manager to develop standardized and reusable bioinformatics pipelines and components.

2. How many pipelines are there currently in nf-core?

There are currently 157 pipelines (103 released, 41 under development, 13 archived)

3. Are there any non-bioinformatic pipelines in nf-core?

Yes, there is a small minority of non-bioinformatics pipelines in nf-core.

4. Let's go back a couple of steps. What is a pipeline and what do we use it for?

A pipeline is a series of processing steps chained together, where the output of one step becomes the input of the next. The steps are defined once, and then the whole chain can be automatically run on any new data.

5. Why do you think nf-core adheres to strict guidelines?

Strict guidelines in a community of many contributors are necessary to ensure reproducibility, consistency, usability, quality, and trustworthiness.

6. What are the main features of nf-core pipelines?

-all pipelines are built with Nextflow and follow the DSL2 syntax standards

-pipelines must be created from the common nf-core template

-all pipelines are open source and released under MIT software license

-software dependencies are containerized with Docker

-functionality is verified by automated CI tests, and pipelines must pass all nf-core pipelines lint tests 

-pipelines are comprehensively documented and developed openly in the nf-core GitHub organisation

Check out the current directory. Next to the outdir you specified, what else has changed?

It also created the log files .nextflow.log, .nextflow.log.1 and .nextflow.log.2, one per run, and the folder .nextflow/ with the history and cache metadata for every run.

```python
# delete the work directory and run the pipeline again using -resume. What did change?
!rm -rf work
!nextflow run nf-core/differentialabundance -r 2.0.0 -profile test,docker --outdir results -resume
```

### What is differential abundance analysis?

Differential abundance analysis tests, for each feature (such as genes, proteins, or metabolites), whether its abundance differs significantly between two conditions, for example treated vs. control. The result is a fold change and a p-value for each feature.

Give the most important plots from the report:

![PCA plot](../day_01/pca2d.png)



PCA plot:
The PCA plot shows the principal component analysis of the 6 samples. PC1 explains 45.3 % and PC2 explains 22.1 % of the variance. PC1 does not clearly separate the treatment groups but seems to reflect variability between individual samples. Along PC2, the mCherry samples tend to lie above the hND6 samples, but the separation is not clear-cut. This suggests that the treatment effect is weak compared to the variability between samples.

![Volcano plot](../day_01/volcano.png)


Volcano plot:
The volcano plot shows the log2 fold change against the adjusted p-value = padj (-log10) for each gene. Most genes are not significant (colored in grey) and lie close to a log2 fold change of 0. Only very few genes (colored in orange) lie at or above the significance threshold of padj = 0.05. Overall, only a small number of genes differ between the treatment groups, which is consistent with the PCA plot, where the groups separate only weakly.
